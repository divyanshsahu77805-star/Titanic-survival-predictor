```python
encoder=OneHotEncoder(drop='first',handle_unknown='ignore',sparse_output=False)
```

```python
y=titanicmodified["Survived"]
```

```python
X=titanicmodified.drop("Survived",axis=1)
```

```python
titanic_encoded=encoder.fit_transform(X)
titanic_encoded.shape
```

```python
import torch
import torch.nn as nn
```

```python
class NeuralNetwork(nn.Module):
    def __init__(self, n_parameters):
        super().__init__()

        self.hidden = nn.Linear(n_parameters, 16)
        self.relu = nn.ReLU()
        self.hidden1=nn.Linear(16,32)
        self.output = nn.Linear(32, 2)

    def forward(self, x):
        x=self.hidden(x)
        x=self.relu(x)
        x=self.hidden1(x)
        x=self.relu(x)
        x=self.output(x)
        return x
```

```python
criterion=nn.CrossEntropyLoss()
model=NeuralNetwork(n_parameters=264)
optimizer=torch.optim.SGD(model.parameters(),lr=0.01)
```

```python
def train_bgd(model, train_loader ,optimizer, criterion, n_epochs=100):
    model.train()
    for X_train, y_train in train_loader:
        for epoch in range(n_epochs):
            optimizer.zero_grad()

            outputs=model(X_train)
            loss=criterion(outputs,y_train)
            loss.backward()
            optimizer.step()

            if (epoch + 1) % 10 == 0:
                print(f"Epoch [{epoch+1}/{n_epochs}], Loss: {loss.item():.4f}")
```

```python
X_tensor = torch.tensor(titanic_encoded, dtype=torch.float32)
y_tensor=torch.tensor(y)
```

```python
from torch.utils.data import TensorDataset, random_split
```

```python
dataset=TensorDataset(X_tensor,y_tensor)
train, valid, test = random_split(dataset, [0.70,0.15,0.15],generator=torch.Generator().manual_seed(42))
```

```python
from torch.utils.data import DataLoader

train_loader = DataLoader(train, batch_size=32, shuffle=True)
```

```python
train_bgd(model, train_loader,optimizer, criterion, 100)
```

```python
def evaluate_loader(model, loader, criterion):
    model.eval()
    total_loss = 0.0
    correct = 0
    total_samples = 0
    
    with torch.no_grad():
        for X_batch, y_batch in loader:
            outputs = model(X_batch)
            loss = criterion(outputs, y_batch)
            
            total_loss += loss.item() * X_batch.size(0)
            _, predicted = torch.max(outputs, 1)
            correct += (predicted == y_batch).sum().item()
            total_samples += y_batch.size(0)
            
    avg_loss = total_loss / total_samples
    accuracy = correct / total_samples
    return avg_loss, accuracy
```

```python
test_loader = DataLoader(test, batch_size=32, shuffle=False)
test_loss, test_accuracy = evaluate_loader(model, test_loader, criterion)
print(f"Test Loss: {test_loss:.4f}, Test Accuracy: {test_accuracy:.4f}")
```
